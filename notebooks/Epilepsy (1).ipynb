#### The Frontal Lobe is responsible for:
* Executive functions (planning, decision-making, problem-solving)
* Motor control (movement, coordination)
* Any seizure detected in Frontal Lobe will directly impact these functions either for short term (1-2 seconds) or for longer durations.

### The Temporal Lobe is responsible for:
* Auditory processing (hearing, sound recognition)
* Memory (short-term and long-term)
* Any seizure detected in Temporal Lobe will impact the Memory or hearing/speech recognition.

### The Parietal lobe is responsible for:
* Sensory processing (touch, spatial awareness)
* Processing of sensory information
* Any seizure detected in Parietal lobe will impact the sensory processing.

### The Occipital lobe is responsible for:
* Visual processing (vision, visual perception)
* Any seizure detected in Occipital lobe will impact the vision.

```python
X_raw = df_data.values          # EEG signals: shape (18117, 178)
y_raw = df_labels['y'].values  # Labels: shape (18117,)
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from scipy.signal import butter, filtfilt
```

```python
def bandpass_filter(data, lowcut=0.5, highcut=45.0, fs=256.0, order=4):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist
    b, a = butter(order, [low, high], btype='band')
    return filtfilt(b, a, data, axis=0)

X_filtered = bandpass_filter(X_raw)
```

#### Nyquist frequency = half of sampling rate = max frequency that can be meaningfully detected

#### -filtfilt() filters the signal forward and backward, no delay or distortion
-Convert cutoff frequencies to the normalized range [0, 1] needed by the butter() function

#### so in short,
A band-pass filter allows only a specific frequency band (here, 1–50 Hz) to pass through and remives frequencies outside this band.

EEG noise includes:

Slow drift (<1 Hz, like baseline wander)

High-frequency noise (>50 Hz, like muscle artifacts or power line interference)

So this filter keeps only the useful EEG bands like delta, theta, alpha, beta, and low gamma.

```python
segment_size = 256
segments = []
segment_labels = []

for i in range(0, len(X_scaled) - segment_size, segment_size):
    segment = X_scaled[i:i+segment_size]
    label = round(np.mean(y_raw[i:i+segment_size]))  # Majority label
    segments.append(segment)
    segment_labels.append(label)

segments = np.array(segments)
segment_labels = np.array(segment_labels)
```

```python
from scipy.stats import skew, kurtosis
```

```python
def extract_statistical_features(segment):
    features = []
    features.extend(np.mean(segment, axis=0))
    features.extend(np.std(segment, axis=0))
    features.extend(skew(segment, axis=0))
    features.extend(kurtosis(segment, axis=0))
    return features
```
